## Cleaning of Date columns

```python
import re

def parse_date(val):
    if pd.isna(val):
        return pd.NaT
    s = str(val).strip()

    # YYYY-MM-DD or YYYY/MM/DD
    if re.match(r'^\d{4}-\d{2}-\d{2}$', s):
        return pd.to_datetime(s, format='%Y-%m-%d', errors='coerce')
    if re.match(r'^\d{4}/\d{2}/\d{2}$', s):
        return pd.to_datetime(s, format='%Y/%m/%d', errors='coerce')

    # DD-MM-YYYY (day > 12, unambiguous) or assumed day-first
    if re.match(r'^\d{2}-\d{2}-\d{4}$', s):
        return pd.to_datetime(s, format='%d-%m-%Y', errors='coerce')

    # MM-DD-YY (2-digit year)
    if re.match(r'^\d{2}-\d{2}-\d{2}$', s):
        return pd.to_datetime(s, format='%m-%d-%y', errors='coerce')

    # MM/DD/YYYY
    if re.match(r'^\d{2}/\d{2}/\d{4}$', s):
        return pd.to_datetime(s, format='%m/%d/%Y', errors='coerce')

    # "Mon DD, YYYY" e.g. Nov 29, 2022
    if re.match(r'^[A-Za-z]{3} \d{1,2}, \d{4}$', s):
        return pd.to_datetime(s, format='%b %d, %Y', errors='coerce')

    # "DD Mon YYYY" e.g. 25 Apr 2022
    if re.match(r'^\d{1,2} [A-Za-z]{3} \d{4}$', s):
        return pd.to_datetime(s, format='%d %b %Y', errors='coerce')

    return pd.NaT  # anything else (e.g. "unknown") becomes NaT

for col in ['order_date', 'ship_date', 'delivery_date']:
    parsed = df[col].apply(parse_date)
    df[col] = parsed.dt.strftime('%d-%m-%Y')  # reformat to dd-mm-yyyy string
    df[col] = pd.to_datetime(df[col], format='%d-%m-%Y', errors='coerce')  # keep as NaT where parsing failed

# Check how many NaT (invalid) values landed in each column
for col in ['order_date', 'ship_date', 'delivery_date']:
    print(f"{col}: {df[col].isna().sum()} invalid/NaT values")

print(df[['order_date', 'ship_date', 'delivery_date']].head(20))
```

***Q1. which customer segments and categories make the most money, and how efficiently ?***

```python
# Keep only orders that actually generated revenue
valid = df[~df["order_status"].isin(["Cancelled", "Returned"])].copy()
print("Rows used:", len(valid), "of", len(df))

# Overall KPIs
total_sales = valid["sales"].sum()
total_profit = valid["profit"].sum()
print(f"Total sales: {total_sales:,.0f}")
print(f"Total profit: {total_profit:,.0f}")
print(f"Overall margin: {total_profit / total_sales:.1%}")

# Performance by customer segment (excluding 'Unknown')
seg = (valid[valid["customer_segment"] != "Unknown"]
       .groupby("customer_segment")
       .agg(orders=("order_id", "count"),
            total_sales=("sales", "sum"),
            total_profit=("profit", "sum"),
            avg_discount=("discount_percent", "mean")))
seg["profit_margin"] = seg["total_profit"] / seg["total_sales"]
print("\nBy segment:")
print(seg.sort_values("total_profit", ascending=False).round(3))

# Performance by category
cat = (valid.groupby("category")
       .agg(orders=("order_id", "count"),
            total_sales=("sales", "sum"),
            total_profit=("profit", "sum")))
cat["profit_margin"] = cat["total_profit"] / cat["total_sales"]
print("\nBy category:")
print(cat.sort_values("total_profit", ascending=False).round(3))
```

* Overall: about 50.4M in sales and 15.0M in profit, a 29.7% margin, from 38,766 valid orders.
* Segments: all five sit between 29.5% and 29.9%. Consumer has roughly double the sales and profit of the others, but that's because it has about twice as many orders (12,449 vs. about 6,300). Its margin is the same as the others, and average discount is also almost identical (14.1% to 14.8%). 
* Categories: margins range only from 29.0% (Furniture) to 30.4% (Clothing). Furniture earns the most profit in currency terms simply because it has the highest sales. 
* The "Unknown" row: it shows 0 orders because those rows were filtered out, so we can ignore it.

**Since segment and category are nearly identical, they can't explain why about 11% of orders lose money. The cause must be something that varies more across orders, most likely discount level and shipping cost.**

***Q4. Does shipping mode affect shipping cost, delivery speed, or profitability?****

```python
# Keep only orders that actually generated revenue
valid = df[~df["order_status"].isin(["Cancelled", "Returned"])].copy()

# 1) Discount level vs profitability
disc = valid.groupby("discount_percent").agg(
    orders=("order_id", "count"),
    total_sales=("sales", "sum"),
    total_profit=("profit", "sum"),
    loss_orders=("profit", lambda x: (x < 0).mean())
)
disc["profit_margin"] = disc["total_profit"] / disc["total_sales"]
print("By discount level:")
print(disc.round(3))

# 2) Shipping mode: cost, speed, margin
ship = valid.groupby("shipping_mode").agg(
    orders=("order_id", "count"),
    avg_shipping_cost=("shipping_cost", "mean"),
    avg_processing_days=("processing_days", "mean"),
    avg_delivery_days=("delivery_days", "mean"),
    total_sales=("sales", "sum"),
    total_profit=("profit", "sum"),
    loss_orders=("profit", lambda x: (x < 0).mean())
)
ship["profit_margin"] = ship["total_profit"] / ship["total_sales"]
print("\nBy shipping mode:")
print(ship.round(3))


# Question: How do discounts affect margin and the share of losing orders?
valid = df[~df["order_status"].isin(["Cancelled", "Returned"])].copy()

disc = valid.groupby("discount_percent").agg(
    total_sales=("sales", "sum"),
    total_profit=("profit", "sum"),
    loss_share=("profit", lambda x: (x < 0).mean())
)

disc["Margin"] = disc["total_profit"] / disc["total_sales"]

result = disc[["Margin", "loss_share"]].copy()
result.index = (result.index * 100).round(0).astype(int).astype(str) + "%"
result.index.name = "Discount"
result.columns = ["Margin", "Share of orders losing money"]

# Show as percentages
result = result.map(lambda v: f"{v:.1%}")
display(result)
```

* Margin falls steadily as the discount rises, roughly 3 to 4 points for every 5 points of discount.
* Up to 20%, no order loses money. At 30%, one in four does. At 50%, three in four do.
* The 50% discount group (4,300 orders, about 3.2M in sales) has a negative total profit of about -637K. Every other group is profitable.
* Zero-discount orders are the biggest group (12,923) and earn about 7.8M profit, over half of all profit.
* Average shipping cost is about 59 to 60 for every mode.
* Average delivery time is about 7 days for every mode, and processing takes about 4.5 days.
* Margin is 29% to 30% and the loss rate is about 11% for every mode.
* So shipping mode is not effective column in this datset.

**This suggests capping discounts at around 20%, though the 30% tier is still profitable overall (14.1% margin) and needs a closer look first.**

***Q5. why do some 30% orders lose money and others don't?***

```python
# Shipping cost as a share of sales
valid["ship_pct_sales"] = valid["shipping_cost"] / valid["sales"]

# Order size buckets (quartiles of sales)
valid["order_size"] = pd.qcut(valid["sales"], 4,
                              labels=["Small", "Medium", "Large", "Very Large"])

# Loss rate by discount and order size
loss = (valid[valid["discount_percent"] >= 0.30]
        .pivot_table(index="discount_percent", columns="order_size",
                     values="profit", aggfunc=lambda x: (x < 0).mean(),
                     observed=True))
print("Share of loss-making orders (discount 30%+ only):")
print(loss.round(3))

# Loss orders vs profitable orders
print("\nLoss vs profitable orders:")
print(valid.groupby(valid["profit"] < 0)[
    ["sales", "quantity", "discount_percent", "shipping_cost", "ship_pct_sales"]
].mean().round(3))
```

* At a 30% discount, about 25% of orders lose money in every order size (Small to Very Large), so order size doesn't explain the losses.
* At a 50% discount, about 74% of orders lose money in every order size, again with almost no difference between sizes.
* Loss-making orders have an average discount of 44.9%, against 10.7% for profitable orders. Discount is the biggest difference between the two groups.
* Average shipping cost is the same for both groups (about 59.7 vs 59.8), so shipping cost is not the cause.
* Average quantity is the same for both groups (about 25.5 vs 25.4), so quantity is not the cause.
* Loss-making orders have lower average sales (828 vs 1,358). This is a result of the deep discounts, not a cause.
* Shipping takes 46.8% of sales on loss-making orders vs 27.7% on profitable ones. This is because their sales are lower, not because shipping costs more.

**Conclusion: deep discounts (30% and above) are what push orders into losses, and order size, shipping cost, and quantity make no difference.**

```python
import matplotlib.pyplot as plt

valid = df[~df["order_status"].isin(["Cancelled", "Returned"])].copy()
fig, ax = plt.subplots(2, 2, figsize=(14, 10))

# 1) Margin by discount level
d = valid.groupby("discount_percent").agg(s=("sales", "sum"), p=("profit", "sum"))
d["margin"] = d["p"] / d["s"] * 100
ax[0, 0].bar((d.index * 100).astype(int).astype(str) + "%", d["margin"],
             color=["#2a9d8f" if m > 0 else "#e76f51" for m in d["margin"]])
ax[0, 0].set_title("Profit margin by discount level")
ax[0, 0].set_ylabel("Margin (%)")
ax[0, 0].axhline(0, color="black", linewidth=0.8)

# 2) Share of losing orders by discount level
l = valid.groupby("discount_percent")["profit"].apply(lambda x: (x < 0).mean() * 100)
ax[0, 1].bar((l.index * 100).astype(int).astype(str) + "%", l, color="#e76f51")
ax[0, 1].set_title("Share of orders losing money by discount level")
ax[0, 1].set_ylabel("Losing orders (%)")

# 3) Loss share by cost ratio (30% discount orders)
h = valid[valid["discount_percent"] == 0.30].copy()
h["cost_ratio"] = h["cost"] / h["unit_price"]
h["is_loss"] = h["profit"] < 0
bins = pd.cut(h["cost_ratio"], [0, 0.48, 0.56, 0.64, 0.72, 1])
c = h.groupby(bins, observed=True)["is_loss"].mean() * 100
ax[1, 0].bar(c.index.astype(str), c, color="#f4a261")
ax[1, 0].set_title("30% discount: losing orders by cost / price ratio")
ax[1, 0].set_ylabel("Losing orders (%)")
ax[1, 0].tick_params(axis="x", rotation=20)

# 4) Monthly sales and profit trend
m = valid.groupby("order_month").agg(sales=("sales", "sum"), profit=("profit", "sum"))
ax[1, 1].plot(m.index, m["sales"], label="Sales")
ax[1, 1].plot(m.index, m["profit"], label="Profit")
ax[1, 1].set_title("Monthly sales and profit")
ax[1, 1].set_xticks(range(0, len(m), 6))
ax[1, 1].set_xticklabels(m.index[::6], rotation=45)
ax[1, 1].legend()

plt.tight_layout()
plt.show()
```

**Chart 1: Margin by discount level**

* Margin falls steadily as the discount rises, from about 40% at 0% discount to 14% at 30%.
* At a 50% discount, the margin is negative (about -20%), so that group loses money overall.

**Chart 2: Share of orders losing money**

* No order loses money at discounts of 0% to 20%.
* Losses start at 30% (about 25% of orders) and become the norm at 50% (about 74%).
* The safe upper limit for discounts in this data is about 20%.

**Chart 3: Cost / price ratio (30% discount orders)**

* Whether a 30% discount order loses money depends on how expensive the product is relative to its price.
* Products costing up to about 64% of their price never lose money, and products above 72% always do.
* Discounts should be set per product based on its cost, not as one flat percentage for everything.

**Chart 4: Monthly sales and profit**

* The business is stable over the four years. Yearly sales stay between 12.5M and 12.7M, with no growth or decline.

# Final Summary: Supply Chain Profitability Analysis

## Objective
Understand what drives profit and loss in the retail supply chain data, and identify where the business can improve.

## Data
- 50,000 orders from January 2021 to December 2024, with 35 columns.
- Cleaned: location fields standardized, invalid or missing values set to "Unknown", dates converted, and new features created (processing days, delivery days, profit margin).
- Cancelled and Returned orders (22.5% of orders) were excluded from the profit analysis, leaving 38,766 completed orders.

## Key Findings
1. **Overall performance is healthy and stable.** Completed orders generated about 50.4M in sales and 15.0M in profit, a 29.7% margin. Yearly sales stayed between 12.5M and 12.7M from 2021 to 2024, with no strong seasonality.
2. **Customer segment and category barely matter.** All five segments have margins of 29.5% to 29.9%, and all six categories 29.0% to 30.4%. Consumer earns the most profit only because it has about twice as many orders.
3. **Discounts are the main driver of profit and loss.** Margin falls from 39.8% at no discount to 14.1% at 30% and -19.7% at 50%.
4. **Losses begin at a 30% discount.** No orders lose money at discounts up to 20%. At 30%, about 25% of orders lose money, and at 50%, about 74% do. The 50% discount group loses about 637K in total.
5. **Order size, shipping cost, and quantity do not explain the losses.** Loss rates are the same across small and large orders, and shipping cost and quantity are identical for loss-making and profitable orders. Only the discount differs (45% vs 11% on average).
6. **Whether a discounted order loses money depends on the product's cost.** At a 30% discount, orders with a cost of up to about 64% of the list price never lose money, while orders above 72% always do.
7. **Shipping mode shows no effect.** Cost (about 60), delivery time (about 7 days), margin (about 30%), and loss rate are nearly identical across all five modes.
8. **Cancellations and returns are not explained by any column.** About 22.5% of orders are cancelled or returned, and the rate is similar across warehouses, segments, categories, shipping modes, payment methods, and discount levels.

## Recommendations
- **Cap discounts at about 20%.** This is the highest level at which no orders lost money.
- **Set discounts per product based on its cost**, not as one flat percentage. Products with a high cost relative to price should get smaller discounts.
- **Review or stop 50% discounts**, which lose money overall.
- **Record the reason for each cancellation or return**, since no existing column explains the 22.5% rate.

## Limitations
- Shipping mode has no effect on delivery time, which is unusual and may mean the column is not reliable.
- Country, state, and city do not always agree with each other, so only country-level location analysis is safe.
- Some "Unknown" values remain (for example in customer segment, payment method, and brand).
- `order_id` is not unique, so it was not used as a key.
- The analysis shows relationships in the data, not proof of cause and effect.
